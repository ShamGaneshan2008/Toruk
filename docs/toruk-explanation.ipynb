## Does Toruk remember exact bytes or only the gist?

The conversational selector retains original text turns. It has no persistent compressed gist store. Exact source bytes are available only if those turns survive selection and prompt fitting. Availability does not guarantee the decoder copies them correctly.

In the development diagnostic, the source motto was `Pause:  two steps!` (two spaces). The new envelope produced `"Pause: two steps!"`: an extra pair of quotes and a missing space. That fails exact recall even though the meaning is close. Source SHA256 and turn IDs are saved separately from the generated answer.

A gist answer can use different words, but must preserve all required meaning. A claim that a compressed vector can reconstruct arbitrary codes perfectly is unsupported. This experiment measures gist answer correctness from selected source text; it does not establish which learned compression is sufficient.


## What happens when 'it' could refer to two things?

The desired answer is a clarification such as 'Which project do you mean?' The current model can still guess even with both candidate records retained. In the development diagnostic, the old selector's answer chose Cobalt's code; the envelope chose Saffron's. Neither resolved the ambiguity. A generic instruction to clarify is not a reliable ambiguity detector.

With one clear project, some long-gap pronoun questions succeed. Those examples do not establish broad reference resolution. Shared weather distractions contain no task-bearing entity. More realistic competing topics may be harder.


## What happens when a correction changes one fact but not another?

Changing a launch date should leave the code intact. The intervention keeps chronological source text instead of replacing an entire entity with a summary. It does not implement field-level memory updates. The decoder must interpret which fact changed.

Development code correction improved from `DV-OLD` to `DV:4_q!`; the unrelated-date correction still left the code answer correct. A different held-out correction failed. Correctness on an example does not guarantee reliable revision or preservation.


## When should Toruk say it does not know?

When evidence is absent, no longer retained, contradictory without a clear correction, or insufficient to identify a referent. Clarification is useful when the user can distinguish candidates. The decoder is still responsible for doing this; the experimental envelope is not an enforced abstention policy.

The frozen lexical uncertainty rubric missed one honest 'I don't have enough context' response. Raw scores remain unchanged and that rubric failure is reported separately. A metric failure and an unsupported factual assertion are different observations.

## Does a smaller prompt mean more efficient memory?

No. Full-history tokenization, selector scans, repeated fitting, backbone prefill, decoding, state bytes and latency all count. This study records them and makes no total-efficiency claim. See the [results and limits](research/recall-results-20261104.md), [frozen protocol](research/recall-v2-protocol.md), and raw artifacts for actual comparisons.


## Strict-template result

On 12 fresh held-out worlds, lexical, old Toruk and the new source envelope each score 4/12 under the frozen rubric, versus 0/12 recent context. Each selection method scores exact 2/6, gist 1/2, clear references 1/2, ambiguity 0/1 and missing evidence 0/1. Strict development regresses 7/12 to 6/12.

The rubric misses some honest uncertainty and accepts an old gist answer with an invented 24-hour retention policy. Raw answers and separate posthoc assistant annotations are preserved; the primary scores are unchanged. New prompt tokens increase from 1,816 to 2,552 over the held-out set. No general memory or total-efficiency win is established.
